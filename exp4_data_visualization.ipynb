## Step 1: Import libraries and set a common plot style

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="colorblind", font_scale=1.1)
```

## Step 2: Load the data and check it

```python
df = pd.read_csv("clinical_trial_data_viz.csv")
visits = pd.read_csv("clinical_visits.csv")

print("Patients:", df.shape)
print("Visits  :", visits.shape)
print()
print("Missing values:")
print(df.isna().sum()[df.isna().sum() > 0])
df.head()
```

## Plot 1: Histogram – Age distribution

```python
plt.figure(figsize=(7, 4.5))
sns.histplot(data=df, x="AGE", bins=15)
plt.title("Age Distribution of Patients")
plt.xlabel("Age (years)")
plt.ylabel("Number of patients")
plt.tight_layout()
plt.savefig("plot01_age_histogram.png", dpi=150)
plt.show()
```

## Plot 2: Histogram – BMI distribution

```python
plt.figure(figsize=(7, 4.5))
sns.histplot(data=df, x="BMI", bins=15)
plt.title("BMI Distribution of Patients")
plt.xlabel("BMI (kg/m²)")
plt.ylabel("Number of patients")
plt.tight_layout()
plt.savefig("plot02_bmi_histogram.png", dpi=150)
plt.show()
```

## Plot 3: Bar chart – Gender distribution

```python
plt.figure(figsize=(5.5, 4.5))
sns.countplot(data=df, x="SEX")
plt.title("Gender Distribution")
plt.xlabel("Sex (M = male, F = female)")
plt.ylabel("Number of patients")
plt.tight_layout()
plt.savefig("plot03_gender_bar.png", dpi=150)
plt.show()
```

## Plot 4: Bar chart – Race distribution

```python
plt.figure(figsize=(7, 4.5))
sns.countplot(data=df, x="RACE", order=df["RACE"].value_counts().index)
plt.title("Race Distribution")
plt.xlabel("Race")
plt.ylabel("Number of patients")
plt.tight_layout()
plt.savefig("plot04_race_bar.png", dpi=150)
plt.show()
```

## Plot 5: Box plot – BMI by age group

```python
plt.figure(figsize=(7, 4.5))
sns.boxplot(data=df, x="AGE_GROUP", y="BMI", order=["<40", "40-59", "60+"])
plt.title("BMI by Age Group")
plt.xlabel("Age group (years)")
plt.ylabel("BMI (kg/m²)")
plt.tight_layout()
plt.savefig("plot05_bmi_by_age_group.png", dpi=150)
plt.show()
```

## Plot 6: Box plot – Blood pressure by treatment arm
Systolic and diastolic pressure are drawn side by side so the arms can be compared on the same scale.

```python
arm_order = ["Placebo", "Drug 10mg", "Drug 20mg"]

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5), sharey=True)

sns.boxplot(data=df, x="ARM", y="SYSBP", order=arm_order, ax=axes[0])
axes[0].set_title("Systolic BP by Treatment Arm")
axes[0].set_xlabel("Treatment arm")
axes[0].set_ylabel("Blood pressure (mmHg)")

sns.boxplot(data=df, x="ARM", y="DIABP", order=arm_order, ax=axes[1])
axes[1].set_title("Diastolic BP by Treatment Arm")
axes[1].set_xlabel("Treatment arm")
axes[1].set_ylabel("")

plt.tight_layout()
plt.savefig("plot06_bp_by_arm.png", dpi=150)
plt.show()
```

## Plot 7: Box plot – Glucose by risk category

```python
plt.figure(figsize=(7, 4.5))
sns.boxplot(data=df, x="RISK_CATEGORY", y="GLUCOSE_MGDL", order=["Low", "Medium", "High"])
plt.title("Glucose by Risk Category")
plt.xlabel("Risk category")
plt.ylabel("Glucose (mg/dL)")
plt.tight_layout()
plt.savefig("plot07_glucose_by_risk.png", dpi=150)
plt.show()
```

## Plot 8: Scatter plot – Height vs weight

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="HEIGHT_CM", y="WEIGHT_KG", hue="SEX")
plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.legend(title="Sex")
plt.tight_layout()
plt.savefig("plot08_height_vs_weight.png", dpi=150)
plt.show()
```

## Plot 9: Scatter plot – Systolic vs diastolic blood pressure

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="SYSBP", y="DIABP", hue="ARM", hue_order=arm_order)
plt.title("Systolic vs Diastolic Blood Pressure")
plt.xlabel("Systolic BP (mmHg)")
plt.ylabel("Diastolic BP (mmHg)")
plt.legend(title="Treatment arm")
plt.tight_layout()
plt.savefig("plot09_sys_vs_dia.png", dpi=150)
plt.show()
```

## Plot 10: Line plot – Mean weight over study weeks
Each line is the average weight of one treatment arm at each visit.

```python
mean_weight = visits.groupby(["WEEK", "ARM"])["WEIGHT_KG"].mean().reset_index()

plt.figure(figsize=(7.5, 5))
sns.lineplot(data=mean_weight, x="WEEK", y="WEIGHT_KG", hue="ARM",
             hue_order=arm_order, marker="o")
plt.title("Mean Weight Over Study Weeks")
plt.xlabel("Study week")
plt.ylabel("Mean weight (kg)")
plt.legend(title="Treatment arm")
plt.tight_layout()
plt.savefig("plot10_weight_over_time.png", dpi=150)
plt.show()
```

## Step 3: Numbers behind the plots (helps you write the observations)

```python
print("Age   : mean %.1f, median %.1f, range %d-%d" % (df.AGE.mean(), df.AGE.median(), df.AGE.min(), df.AGE.max()))
print("BMI   : mean %.1f, median %.1f, skewness %.2f" % (df.BMI.mean(), df.BMI.median(), df.BMI.skew()))
print()
print(df["SEX"].value_counts(), "\n")
print(df["RACE"].value_counts(), "\n")
print("Median SYSBP by arm:");  print(df.groupby("ARM")["SYSBP"].median().round(1), "\n")
print("Median glucose by risk:"); print(df.groupby("RISK_CATEGORY")["GLUCOSE_MGDL"].median().round(1), "\n")
print("Height-weight correlation: %.2f" % df["HEIGHT_CM"].corr(df["WEIGHT_KG"]))
print("SYSBP-DIABP correlation  : %.2f" % df["SYSBP"].corr(df["DIABP"]))
```

## Step 4: Outliers and data-quality checks (rows 14 and 15 of the sheet)

```python
print("Missing glucose values:", df["GLUCOSE_MGDL"].isna().sum())

# IQR rule: a value is an outlier if it lies 1.5 x IQR beyond the quartiles
for col in ["BMI", "SYSBP", "DIABP", "GLUCOSE_MGDL", "WEIGHT_KG"]:
    q1, q3 = df[col].quantile([0.25, 0.75])
    low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
    outliers = df[(df[col] < low) | (df[col] > high)]
    print(f"{col}: {len(outliers)} outliers ->", outliers["USUBJID"].tolist())

print()
print("Short tall-heavy check (height < 155 cm and weight > 100 kg):")
df[(df["HEIGHT_CM"] < 155) & (df["WEIGHT_KG"] > 100)][["USUBJID", "HEIGHT_CM", "WEIGHT_KG", "BMI"]]
```
